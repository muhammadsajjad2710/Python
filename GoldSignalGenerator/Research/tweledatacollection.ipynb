```python
import requests
import pandas as pd
from datetime import datetime

# --- API Configuration ---
api_key = "8f5afc8625b4427683cdfc5af6c60f3d"
base_url = "https://api.twelvedata.com/time_series"

# --- Parameters ---
symbol = "XAU/USD"  # Gold price in USD
interval = "1h"     # Hourly data
start_date = "2022-04-08"
end_date = "2024-04-08"

# --- API Request ---
params = {
    "symbol": symbol,
    "interval": interval,
    "start_date": start_date,
    "end_date": end_date,
    "apikey": api_key,
    "format": "JSON",
    "outputsize": 5000  # get as many data points as possible
}

response = requests.get(base_url, params=params)

# --- Check Response ---
if response.status_code == 200:
    data = response.json()
    if "values" in data:
        df2 = pd.DataFrame(data["values"])
        df2["datetime"] = pd.to_datetime(df2["datetime"])
        df2 = df2.rename(columns={
            "datetime": "UTC",
            "open": "Open",
            "high": "High",
            "low": "Low",
            "close": "Close",
            "volume": "Volume"
        })

        # Define the list of columns to select, conditionally including 'Volume'
        cols_to_select = ["UTC", "Open", "High", "Low", "Close"]
        if "Volume" in df2.columns:
            cols_to_select.append("Volume")

        df2 = df2[cols_to_select]
        df2 = df2.sort_values("UTC")
        print(df2.head())
    else:
        print("No 'values' field in response:", data)
else:
    print("Error:", response.status_code, response.text)
```

combined_df is final data frame . it is the combination of three data frames df, df1 and df2.

```python
import pandas as pd

# --- Ensure UTC is a regular column ---
for d in [df, df1, df2]:
    if d.index.name == "UTC":
        d.reset_index(inplace=True)
    d["UTC"] = pd.to_datetime(d["UTC"], errors="coerce")

    # Normalize all to UTC without tz-awareness
    if d["UTC"].dt.tz is not None:
        d["UTC"] = d["UTC"].dt.tz_convert(None)

# --- Combine DataFrames ---
combined_df = pd.concat([df, df1, df2], ignore_index=True)

# Drop missing UTC rows
combined_df = combined_df.dropna(subset=["UTC"])

# Force timezone-naive consistency again
combined_df["UTC"] = pd.to_datetime(combined_df["UTC"], utc=False)

# Sort chronologically
combined_df = combined_df.sort_values("UTC").reset_index(drop=True)

# Remove duplicates (in case of overlap)
combined_df = combined_df.drop_duplicates(subset=["UTC"], keep="last")

print(combined_df.head())
print(combined_df.tail())
print(f"\n✅ Total rows after cleaning: {len(combined_df)}")

# --- Save final combined data ---
combined_df.to_csv("gold_hourly_combined.csv", index=False)
print("\n💾 Saved to gold_hourly_combined.csv")
```
